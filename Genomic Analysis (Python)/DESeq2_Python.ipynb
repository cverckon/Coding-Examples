```python
import pandas as pd
from pydeseq2.dds import DeseqDataSet
from pydeseq2.ds import DeseqStats

```

```python
# counts created as dataframe export of airway dataset from R passed through assay()
# meta created as dataframe export of airway dataset from R passed through colData()
counts = pd.read_csv("airway_counts.csv").T
meta = pd.read_csv("airway_meta.csv", index_col = 4)
```

```python
keep = (counts > 10).sum(axis=0) >= 4
counts_filtered = counts.loc[:, keep]

dds = DeseqDataSet(
    counts=counts_filtered,
    metadata=meta,
    design="~ cell + dex",
    ref_level = "untrt"
)

dds.deseq2()

ds_trt = DeseqStats(dds, contrast = ["dex", "trt", "untrt"])
```
